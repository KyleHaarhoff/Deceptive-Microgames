```python
ave_res = pd.read_csv("DQN/The_Resistance/ave-rewards-run_1.csv")
ave_res = ave_res[['spy1', 'agent1']]


learning_stats = pd.read_csv("DQN/The_Resistance/game-results-run_1.csv")
learning_stats["Percentage_of_spy_to_spy_yes"] = learning_stats["total_spy_to_spy_votes_yes"] / learning_stats["total_spy_to_spy_votes"]
learning_stats["Percentage_of_sabotaged_infested_missions"] = learning_stats["total_infested_missions_sabotaged"] / learning_stats["total_infested_missions"]
learning_stats["Percentage_of_agents_accepting_others"] = learning_stats["total_agent_not_in_vote_yes"] / learning_stats["total_agent_not_in_vote"]


learning_stats["agents_win"] = (learning_stats["results"] == 1).astype(int)
learning_stats["spies_win"] = (learning_stats["results"] == -1).astype(int)
```

```python
window = 1000  


fig, axs = plt.subplots(3, 2, figsize=(12, 18))  

window = 1000  

for i in range(5):
    learning_stats = pd.read_csv(f"DQN/The_Resistance/game-results-run_{i+1}.csv")
    learning_stats["agents_win"] = (learning_stats["results"] == 1).astype(int)
    learning_stats["spies_win"] = (learning_stats["results"] == -1).astype(int)
    rolling = learning_stats.rolling(window=window).mean()
    axs[i//2][i%2].plot(rolling["spies_win"], color="red")
    axs[i//2][i%2].plot(rolling["agents_win"], color="blue")
    
    axs[i//2][i%2].set_xlabel("Episode")
    axs[i//2][i%2].set_ylabel(f"Win Proportion")
    
    axs[i//2][i%2].set_title(f"Average Win Proportion per Episode for Run {i+1}")
    axs[i//2][i%2].legend(["Spy", "Agent"])
    axs[i//2][i%2].grid(True)
    
axs[2][1].axis('off')
plt.show()
```
